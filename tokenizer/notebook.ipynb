# GPT-2 Tokenizer
### I. Preliminaries - the Byte-Pair Encoding Algorithm (BPE)

```python
from typing import List, Tuple, Dict

def bpe(input: str) -> Tuple[List[int], Dict[int, tuple]]:
    """encodes a str by increasing its vocab size to represent most frequent pairs of bytes"""
    encoded = input.encode()
    mappings = map(int, encoded)
    tokens = list(mappings)
    top_pair = most_freq_pair(tokens)
    max_token = max(tokens)
    token_to_pair = {}
    while top_pair is not None:
        new_token = max_token + 1
        # assign the new token to the pair it replaces
        token_to_pair[new_token] = top_pair
        max_token += 1
        i = 0
        while i < len(tokens)-1:
            if (tokens[i], tokens[i+1]) == top_pair:
                del tokens[i+1]
                tokens[i] = new_token
            else:
                i += 1
        top_pair = most_freq_pair(tokens)
    return tokens, token_to_pair

def most_freq_pair(elts: list):
    """returns the most frequent pair of elts in the list"""
    pairs = [(elts[i], elts[i+1]) for i in range(len(elts)-1)]
    if not pairs:
        return None
    freq = {}
    for pair in pairs:
        freq[pair] = freq.get(pair, 0) + 1
    most_common = None
    max_count = 0
    for pair, count in freq.items():
        if count > max_count:
            max_count = count
            most_common = pair
    if max_count <= 1:
        return None
    return most_common

sample = "If you've made it this far, you're now a pro at LLM Tokenization! Sadly, you're not exactly done yet because a lot of LLMs outside of OpenAI (e.g. Llama, Mistral) use sentencepiece instead. Primary difference being that sentencepiece runs BPE directly on Unicode code points instead of on UTF-8 encoded bytes. Feel free to explore sentencepiece on your own (good luck, it's not too pretty), and stretch goal if you really experience and suffer from the burden of time, re-write your BPE to be on Unicode code points and match the Llama 2 tokenizer."
print(list(map(int, sample.encode())))
print(bpe(sample))
print(len(sample))
print(len(bpe(sample)))
print(len(set(sample)))
```
